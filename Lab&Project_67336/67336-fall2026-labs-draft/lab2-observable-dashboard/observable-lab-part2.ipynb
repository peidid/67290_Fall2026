<h2>Background: What Is AI-Assisted (Vibe) Coding?</h2>

**AI-assisted coding** (sometimes called *vibe coding*) is the practice of using a large language model (LLM) — such as Google Gemini, ChatGPT, or GitHub Copilot — to help write, debug, or explain code. Instead of writing every line yourself, you *describe* what you want in plain English and the AI generates code for you to review and use.

This is increasingly common in professional software development, data science, and dashboard work. However, AI is not magic:

- It can produce code that **looks correct but has subtle bugs**.
- It may generate code that **does not match your data structure** unless you provide context.
- It can suggest visualizations that are **technically valid but misleading or cluttered**.

A skilled practitioner knows when to *trust* the AI output, when to *fix* it, and when to *reject* it entirely. That is exactly the skill you will practice in this lab.

> **Key concept:** The AI is a *collaborator*, not the author. You are responsible for the quality of the final result.

<h2>Challenge 2: Learning to Prompt Effectively</h2>

The quality of AI output depends heavily on the quality of your prompt. A vague prompt produces vague or incorrect code; a specific, context-rich prompt produces far more useful results.

<h3>Anatomy of a good coding prompt</h3>

A good prompt for code generation typically includes:

1. **Context** – what library/framework/language you are using
2. **Data description** – what the data looks like (field names, types, shape)
3. **Goal** – what you want the code to do
4. **Constraints** – any requirements (e.g., must use `Plot.cell`, must accept a `width` argument)

**Weak prompt** ❌
> *"Write code to show the weather data."*

**Strong prompt** ✅
> *"I am using Observable Framework with the Observable Plot library. My data is stored in a variable called `forecast` which has a `properties.periods` array. Each period has a `startTime` (ISO timestamp string), `temperature` (number in °F), `windSpeed` (string like "10 mph"), and `shortForecast` (string description). Write a JavaScript function called `windSpeedPlot(data, {width} = {})` that returns a `Plot.plot()` showing wind speed over time as a bar chart. Parse the numeric value out of the windSpeed string. Include a y-axis label and a chart title."*

Notice how the strong prompt specifies the library, the exact data structure, the function signature, and the chart type. This leaves the AI much less room to guess.

<h2>Challenge 3: Adding a Wind Speed Chart with AI Assistance</h2>

The NWS forecast data you loaded in Part 1 contains more than just temperature. Each period also has a `windSpeed` field (a string like `"12 mph"`) and a `windDirection` field (a string like `"NW"`). You will use Gemini to help you add a **wind speed bar chart** to your dashboard.

<h3>Challenge 3.1: Craft your prompt</h3>

Using the prompt structure from Challenge 2, write a prompt in Gemini asking it to create a component function called `windSpeedPlot(data, {width} = {})` that:

- Uses `Plot.rectY` or `Plot.barY` to plot wind speed over time
- Parses the numeric part of the `windSpeed` string (e.g., extracts `12` from `"12 mph"`)
- Uses `startTime` on the x-axis
- Displays the chart title and labeled axes
- Returns a `Plot.plot()` object (so it works as a component)

**Write your own prompt** — do not just copy the one from the example above. Then submit it to Gemini.

<h3>Challenge 3.2: Read and evaluate the output</h3>

Before copying *anything*, read the code Gemini generated carefully:

- Does it use `data.properties.periods`, matching your actual data structure?
- Does it correctly extract the number from the wind speed string? (Hint: look for `parseInt()`, `parseFloat()`, or a regex like `/\d+/`.)
- Does the function return a `Plot.plot({...})` object, or does it try to display it directly with `display()`?
- Are there any syntax errors visible at a glance?

If any of these are wrong, **fix the prompt** and ask Gemini again. Iteration is normal and expected.

<h3>Challenge 3.3: Integrate the code</h3>

Once you are satisfied with the generated code:

1. Copy the function from Gemini.
2. Paste it into a new `js` code block in your `weather.md` file.
3. Add a call to `display(windSpeedPlot(forecast))` in a second code block to render it.
4. Check the live preview. Does the chart appear? Does it look correct?

If the chart does not appear or shows an error, read the error message in the browser console (right-click → Inspect → Console) and use that error message in a follow-up prompt to Gemini to help debug it.

<h3>Challenge 3.4: Add it to the dashboard grid</h3>

Once you verify the chart works using `display(windSpeedPlot(forecast))`, **delete that standalone `display()` call** before adding it to the grid below. Otherwise, your chart will render twice on the page.

Update the grid layout in `weather.md` to include `windSpeedPlot` as a fourth card. You can decide whether to add it as a full-width row or alongside an existing chart. Use `resize` just like the other charts:

```
${resize((width) => windSpeedPlot(forecast, {width}))}
```
